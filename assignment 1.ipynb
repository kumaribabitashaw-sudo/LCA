```python
# Student Data Management using Dictionary, Tuple, and List

# Dictionary to store student records
# Key = Roll Number, Value = Tuple of (Name, Branch, Marks)
students = {
    1: ("Amit", "CSE", 85),
    2: ("Priya", "ECE", 90),
    3: ("Rahul", "Mechanical", 75)
}

# List to keep track of all roll numbers
roll_numbers = list(students.keys())

print("----- Initial Student Records -----")
for roll, details in students.items():
    print(f"Roll No: {roll}, Name: {details[0]}, Branch: {details[1]}, Marks: {details[2]}")

# 1. Add a new student record
def add_student(roll, name, branch, marks):
    students[roll] = (name, branch, marks)
    roll_numbers.append(roll)
    print(f"\nStudent with Roll No {roll} added successfully.")

# 2. Delete an existing student record
def delete_student(roll):
    if roll in students:
        del students[roll]
        roll_numbers.remove(roll)
        print(f"\nStudent with Roll No {roll} deleted successfully.")
    else:
        print(f"\nStudent with Roll No {roll} not found.")

# 3. Update the details of a student
def update_student(roll, name, branch, marks):
    if roll in students:
        students[roll] = (name, branch, marks)
        print(f"\nStudent with Roll No {roll} updated successfully.")
    else:
        print(f"\nStudent with Roll No {roll} not found.")

# 4. Display final student records
def display_students():
    print("\n----- Final Student Records -----")
    for roll, details in students.items():
        print(f"Roll No: {roll}, Name: {details[0]}, Branch: {details[1]}, Marks: {details[2]}")

# ---- Performing Operations ----

# Add a new student
add_student(4, "Sneha", "IT", 88)

# Delete an existing student
delete_student(2)

# Update a student's details
update_student(1, "Amit Kumar", "CSE", 95)

# Display final records
display_students()

print("\nAll Roll Numbers:", roll_numbers)
```
